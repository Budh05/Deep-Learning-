Transformation of line in perceptron

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
def perceptron(X, y):
  X = np.insert(X, 0, 1, axis=1) # inserting 1 as bias term coefficient
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(1000):
    j = np.random.randint(0, 100)
    # Compute dot product, then pass to step function
    y_hat = step(np.dot(X[j], weights))
    weights = weights + lr * (y[j] - y_hat) * X[j]
  return weights[0], weights[1:]

intercept_, coef_ = perceptron(X, y)
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
print(coef_)
print(intercept_)
```

```python
m=-(coef_[0]/coef_[1])
b=intercept_/coef_[1]
```

```python
x_input=np.linspace(-3,3,100)
y_input=m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color='red', linewidth=3, label='Decision Boundary')
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', edgecolors='k', s=100)
plt.ylim(-3, 3)
plt.legend()
plt.title('Perceptron Decision Boundary')
plt.show()
```
