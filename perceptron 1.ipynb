```python
from sklearn.model_selection import train_test_split

# We should use only 'cgpa' for X, since 'package' is the target
X = df[['cgpa']]
y = df['package']

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Initialize the model
lr = LinearRegression()

# Train the model
lr.fit(x_train, y_train)

# Make predictions on test data
y_pred = lr.predict(x_test)

# Evaluate the model
print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
print("Slope (Coefficient):", lr.coef_[0])
print("Intercept:", lr.intercept_)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the data points
sns.scatterplot(x=x_train['cgpa'], y=y_train, label='Train Data')

# Plot the linear regression line
plt.plot(x_train['cgpa'], lr.predict(x_train), color='red', linewidth=2, label='Regression Line')

plt.title('Regression Line on Training Data')
plt.xlabel('CGPA')
plt.ylabel('Package')
plt.legend()
plt.show()
```
